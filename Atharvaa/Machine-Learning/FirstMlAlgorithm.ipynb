```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [[150, 0],
            [170, 0],
            [130, 1],
            [120, 1]
            ]
```

```python
labels = ["apple", "apple", "orange", "orange"]
```

```python
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit(features, labels)
```

```python
predection = clf.predict([[150, 1], [34, 1]])
```

```python
print(f"The fruit is {predection[0]}")
```
